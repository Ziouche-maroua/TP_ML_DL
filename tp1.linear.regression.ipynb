# Practical Work 1: Simple Linear Regression

In this notebook, we explore a simple example of **Linear Regression** using Python libraries:
- **NumPy** for numerical calculations
- **Matplotlib** for data visualization
- **Scikit-learn** for machine learning models

The goal is to understand how a regression model learns a relationship between two variables, how we can make predictions, and how to evaluate the model’s performance.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## 1. Generate sample data

We start by creating simple data points.  
- `X` represents the **input (independent variable)**  
- `y` represents the **output (dependent variable)**  
We reshape `X` into a column vector using `.reshape(-1,1)` because scikit-learn expects 2D inputs.


```python
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 5, 7])
```

## 2. Create and train the model

We create a `LinearRegression` model from scikit-learn and train it using `model.fit(X, y)`.  
The model will learn the best line that minimizes the distance between the data points and the line.


```python
model = LinearRegression()
model.fit(X, y)
```

## 3. Display model parameters

After training, the model has two main parameters:
- **Coefficient (a)**: the slope of the line  
- **Intercept (b)**: the value of y when x = 0  

These define the linear equation: **y = a × x + b**


```python
print("Coefficient (a):", model.coef_[0])
print("Intercept (b):", model.intercept_)
```

## 4. Make predictions

We now use the model to predict values of y for new x values (7, 8, and 9).  
This shows how the model can estimate unseen data based on the learned pattern.


```python
X_new = np.array([[7], [8], [9]])
y_pred = model.predict(X_new)
print("\nPredictions for X = 7, 8, 9:", y_pred)
```

## 5. Visualization

We visualize three things:
- **Blue dots** → training data  
- **Red line** → regression line (the model)  
- **Green dots** → predictions for new values  

This helps us understand how the model fits the data and where the new predictions fall.


```python
plt.scatter(X, y, color='blue', label="Training data")
plt.plot(X, model.predict(X), color='red', label="Regression line")
plt.scatter(X_new, y_pred, color='green', label="Predictions")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression Example")
plt.legend()
plt.show()
```

## 6. Evaluate the model

We calculate the **Mean Squared Error (MSE)** to measure how close the model’s predictions are to the actual data.  
A smaller MSE means better performance.

> Note: We compare the model’s predictions on the same training data to check how well it fits.


```python
mse = np.mean((y - model.predict(X)) ** 2)
print("\nMean Squared Error on training data:", mse)
```
