### PART 4 – Email Spam Detection

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import mean_absolute_error
```

## Dataset

```python
X_train = np.array([[0,1],[1,2],[2,3],[3,4],[5,6],[0,0],[4,5]])
y_train = np.array([0,0,1,1,1,0,1])
X_test = np.array([[2,1]])
y_true = np.array([1])
```

## Ploat the training data

```python

plt.scatter(X_train[y_train==0,0], X_train[y_train==0,1], color='green', label='Normal')
plt.scatter(X_train[y_train==1,0], X_train[y_train==1,1], color='red', label='Spam')
plt.scatter(2,1, color='blue', s=100, label='New Email')
plt.xlabel("Links")
plt.ylabel("Spam Words")
plt.legend()
plt.title("Email Spam Training Data")
plt.show()
```

## train,predict and compute MAE

```python
for k in [1,3,5]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    mae = mean_absolute_error(y_true, y_pred)
    print(f"[Email] K={k}, Pred={y_pred[0]}, MAE={mae}")
```
